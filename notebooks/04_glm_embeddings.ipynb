# 04. Genomic Language Model (GLM) Embeddings
This notebook loads the trained BERT-style Genomic Language Model, inspects CBOW-initialized token embeddings ($\mathbf{E}_{CBOW} + \mathbf{P}$), and extracts contextual hidden state representations ($\mathbf{H}$) for RNA transcripts.

```python
import torch
from transformers import BertForMaskedLM, AutoTokenizer
from pathlib import Path

PROJECT_ROOT = Path('.').resolve().parents[0]
MODEL_PATH = PROJECT_ROOT / 'models' / 'glm' / 'final_model'
TOKENIZER_PATH = PROJECT_ROOT / 'models' / 'tokenizers' / 'glm_bpe_tokenizer.json'

print(f'Looking for GLM checkpoint at: {MODEL_PATH}')
```

## Loading Model Checkpoints and Inspecting Embeddings

```python
if MODEL_PATH.exists() and TOKENIZER_PATH.exists():
    model = BertForMaskedLM.from_pretrained(str(MODEL_PATH))
    model.eval()
    
    print('Model Architecture Configuration:')
    print(model.config)
    
    # Inspect embedding layer weights (E_CBOW + P)
    embeddings = model.bert.embeddings.word_embeddings.weight
    print(f'Word embedding matrix shape: {embeddings.shape}')
else:
    print('GLM model checkpoint or tokenizer not found. Run workflow 05 first to train the GLM.')
```
