## Step 3: Spatial Kernel & Interaction Simulation

```python
# Generate synthetic tissue spatial coordinates for demonstration
np.random.seed(42)
num_spots = 100
coords = np.random.uniform(0, 1000, size=(num_spots, 2))

# Compute Gaussian spatial kernel K_S
sigma_S = 150.0
K_S = compute_spatial_kernel(coords, sigma_S=sigma_S, sparsity_threshold=0.05)

print(f'Spatial coordinate shape: {coords.shape}')
print(f'Computed Spatial Kernel shape: {K_S.shape}')
print(f'Sparsified non-zero edges: {np.count_nonzero(K_S)}')
print('Tutorial execution complete!')
```
