## 3. Spatial Tissue Topology & Gaussian Kernel ($\mathcal{K}_{ij}^S$)
Computing physical spatial distance decay between tissue spots:
$$K_{ij}^S = \exp\left(-\frac{\|\mathbf{s}_i - \mathbf{s}_j\|^2}{2\sigma_S^2}\right)$$

```python
num_spots = 50
coords = np.random.uniform(0, 500, size=(num_spots, 2))
sigma_S = 100.0

K_S = compute_spatial_kernel(coords, sigma_S=sigma_S, sparsity_threshold=0.05)
print(f'Spatial coordinate matrix shape: {coords.shape}')
print(f'Sparsified Spatial Kernel shape: {K_S.shape}')
print(f'Total active spatial interaction edges: {np.count_nonzero(K_S)}')
```

## 4. Graph Diffusion Signal Propagation ($\mathbf{F}(t) = e^{-t\mathcal{L}_E}\mathbf{f}_0$)
Modeling exerkine signaling across tissue spots via Graph Laplacian heat diffusion.

```python
from scipy.sparse import csr_matrix, diags

# Construct mock adjacency and Laplacian L_E = D - W
W_mock = csr_matrix((np.ones(num_spots), (np.arange(num_spots), (np.arange(num_spots) + 1) % num_spots)), shape=(num_spots, num_spots))
D_mock = diags(np.array(W_mock.sum(axis=1)).flatten())
L_E = D_mock - W_mock

f_0 = np.random.uniform(0, 1, size=num_spots)
print(f'Graph Laplacian shape: {L_E.shape}')
print(f'Initial signal vector f_0 shape: {f_0.shape}')
print('Core concepts walkthrough successfully demonstrated!')
```
