# 02. Single-Cell Quality Control & Annotation
This notebook processes single-cell transcriptomics data ($\mathbf{X}$) for the EXERKINEMAP framework, performing quality filtering, normalization, highly variable gene extraction, and UMAP dimensionality reduction.

```python
import scanpy as sc
import matplotlib.pyplot as plt
from pathlib import Path

PROJECT_ROOT = Path('.').resolve().parents[0]
PROCESSED_SC_DIR = PROJECT_ROOT / 'data' / 'processed' / 'anndata'

# Set up plotting parameters
sc.settings.verbosity = 3
sc.settings.set_figure_params(dpi=80, facecolor='white')
```

## Loading Processed AnnData Matrix

```python
h5ad_path = PROCESSED_SC_DIR / 'exerkinemap_sc_processed.h5ad'
if h5ad_path.exists():
    adata = sc.read_h5ad(h5ad_path)
    print(adata)
else:
    print(f'Processed h5ad not found at {h5ad_path}. Run workflow 02 first.')
```

## Dimensionality Reduction and UMAP Visualization
Projecting cellular states into low-dimensional space for visualization according to Section 16.

```python
if 'adata' in locals() and adata.n_obs > 0:
    # Compute neighborhood graph and UMAP if not already computed
    if 'neighbors' not in adata.uns:
        sc.pp.neighbors(adata, n_neighbors=15, n_pcs=30)
    if 'umap' not in adata.obsm:
        sc.tl.umap(adata)
        
    # Plot UMAP colored by cell type or clusters if available
    color_key = 'cell_type' if 'cell_type' in adata.obs else adata.obs_keys()[0]
    sc.pl.umap(adata, color=[color_key], title='EXERKINEMAP Cellular UMAP Embedding')
else:
    print('AnnData object is empty or uninitialized.')
```
