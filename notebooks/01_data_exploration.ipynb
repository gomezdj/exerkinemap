# 01. Data Exploration & Raw Ingestion
This notebook explores the raw single-cell transcripts, metadata CSV files, and spatial omics data downloaded for the EXERKINEMAP framework.

```python
import pandas as pd
import scanpy as sc
from pathlib import Path

# Resolve project root directory
PROJECT_ROOT = Path('.').resolve().parents[0]
print(f'Project Root: {PROJECT_ROOT}')

RAW_DIR = PROJECT_ROOT / 'data' / 'raw'
print(f'Raw Data Directory: {RAW_DIR}')
```

## Inspecting Single-Cell Metadata

```python
meta_path = RAW_DIR / 'metadata' / 'motrpac_sc_metadata.csv'
if meta_path.exists():
    meta_df = pd.read_csv(meta_path)
    print(f'Metadata shape: {meta_df.shape}')
    display(meta_df.head())
else:
    print('Metadata CSV not found. Ensure raw ingestion workflow 01 has been run.')
```

## Inspecting Raw AnnData (.h5ad) Objects

```python
h5ad_path = RAW_DIR / 'single_cell' / 'motrpac_sc_transcriptomics.h5ad'
if h5ad_path.exists():
    adata = sc.read_h5ad(h5ad_path, backed='r')
    print(adata)
else:
    print('Raw AnnData file not found. Ensure raw data staging is complete.')
```
